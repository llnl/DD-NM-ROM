```python
import os
import json
import numpy as np
```

```python
elements = ["interior", "port"]
prefix = "/g/g92/zanardi1/Workspace/Codes/DD-NM-ROM/run/unsteady/dset.2by2/figures/"
tag = "interior_ld_24_rnz_10_port_ld_10_rnz_8"
```

```python
errors = []
for icase in os.scandir(prefix+tag):
  if ((icase.is_dir()) and ("case" in icase.name)):
    filename = icase.path+"/rom/srpc/stats.json"
    try:
      with open(filename, 'r') as file:
        stats = json.load(file)
      index = int(icase.name.split("_")[-1])-1
      errors.append((index, stats["error"]))
    except:
      continue
errors = np.array(errors)
```

```python
i = errors[:,-1].argsort()
errors = errors[i][:,0].astype(int)
print("Best:", errors[:10]+1)
print("Worst:", errors[-10:]+1)
```
